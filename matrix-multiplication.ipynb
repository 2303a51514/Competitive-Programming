```python
import sys
def matrix_chain_multiplication(p, n):
    m = [[0 for _ in range(n + 1)] for _ in range(n + 1)]
    for i in range(1, n + 1):
        m[i][i] = 0
    for L in range(2, n + 1):
        for i in range(1, n - L + 2):
            j = i + L - 1
            m[i][j] = sys.maxsize
            for k in range(i, j):
                cost = (
                    m[i][k] +
                    m[k + 1][j] +
                    p[i - 1] * p[k] * p[j]
                )

                if cost < m[i][j]:
                    m[i][j] = cost

    return m[1][n], m
n = int(input("Enter number of matrices: "))
p = []
print("Enter dimension array values:")
for i in range(n + 1):
    p.append(int(input(f"p[{i}] = ")))
min_cost, dp_table = matrix_chain_multiplication(p, n)
print("\nMinimum number of scalar multiplications:", min_cost)
print("\nDP Table:")
for i in range(1, n + 1):
    print(dp_table[i][1:n + 1])
```
